```python
import pandas as pd

train = pd.read_csv('train.csv')
store = pd.read_csv('store.csv')

print(train.shape)
print(store.shape)
train.head()
```

```python
import os
print(os.getcwd())
print(os.listdir())
```

```python
# Check date range
print("Date range:", train['Date'].min(), "to", train['Date'].max())

# Check for missing values
print("\nMissing values in train:\n", train.isnull().sum())
print("\nMissing values in store:\n", store.isnull().sum())

# Check unique values in key categorical columns
print("\nStateHoliday unique values:", train['StateHoliday'].unique())
print("Open unique values:", train['Open'].unique())

# Look at store metadata
store.head()
```

```python
# Check sales when store is closed
print(train[train['Open'] == 0]['Sales'].describe())

# What fraction of days are stores closed?
print("\n% of days closed:", (train['Open'] == 0).mean() * 100)

# Are there any days where store is open but sales = 0?
print("\nOpen but zero sales:", ((train['Open'] == 1) & (train['Sales'] == 0)).sum())
```

```python
# Drop closed-store days
train_open = train[train['Open'] == 1].copy()

# Inspect the anomalous zero-sales-while-open rows
zero_sales_open = train_open[train_open['Sales'] == 0]
print(zero_sales_open.shape)
zero_sales_open.head(10)
```

```python
# Drop the anomalous zero-sales rows too
train_clean = train_open[train_open['Sales'] > 0].copy()

print("Original:", train.shape)
print("After removing closed days:", train_open.shape)
print("After removing zero-sales anomalies:", train_clean.shape)
```

```python
# Merge store metadata into the cleaned sales data
df = train_clean.merge(store, on='Store', how='left')

print(df.shape)
df.head()
```

```python
# Convert Date to proper datetime
df['Date'] = pd.to_datetime(df['Date'])

# Sort by store and date (critical for lag/rolling features later)
df = df.sort_values(['Store', 'Date']).reset_index(drop=True)

# Extract calendar features
df['Year'] = df['Date'].dt.year
df['Month'] = df['Date'].dt.month
df['Day'] = df['Date'].dt.day
df['WeekOfYear'] = df['Date'].dt.isocalendar().week
df['Quarter'] = df['Date'].dt.quarter

df[['Date', 'Year', 'Month', 'Day', 'WeekOfYear', 'Quarter']].head()
```

```python
# Lag features: sales N days ago, computed within each store group
df['Sales_Lag_1'] = df.groupby('Store')['Sales'].shift(1)
df['Sales_Lag_7'] = df.groupby('Store')['Sales'].shift(7)
df['Sales_Lag_14'] = df.groupby('Store')['Sales'].shift(14)

# Rolling averages: smoothed recent trend, also per store
df['Sales_RollingMean_7'] = df.groupby('Store')['Sales'].transform(lambda x: x.shift(1).rolling(7).mean())
df['Sales_RollingMean_30'] = df.groupby('Store')['Sales'].transform(lambda x: x.shift(1).rolling(30).mean())

df[['Store', 'Date', 'Sales', 'Sales_Lag_1', 'Sales_Lag_7', 'Sales_Lag_14', 'Sales_RollingMean_7', 'Sales_RollingMean_30']].head(15)
```

```python
# Drop rows with NaN in our engineered features (early days per store without enough history)
df_model = df.dropna(subset=['Sales_Lag_1', 'Sales_Lag_7', 'Sales_Lag_14', 'Sales_RollingMean_7', 'Sales_RollingMean_30']).copy()

print("Before dropping NaNs:", df.shape)
print("After dropping NaNs:", df_model.shape)

# Standardize StateHoliday to string (fixes the mixed dtype warning from earlier)
df_model['StateHoliday'] = df_model['StateHoliday'].astype(str)

# Is today a holiday of any kind (state or school)?
df_model['IsHoliday'] = ((df_model['StateHoliday'] != '0') | (df_model['SchoolHoliday'] == 1)).astype(int)

# Competition age in months (how long has competition existed, if known)
df_model['CompetitionOpenMonths'] = (
    (df_model['Year'] - df_model['CompetitionOpenSinceYear']) * 12 +
    (df_model['Month'] - df_model['CompetitionOpenSinceMonth'])
)
df_model['CompetitionOpenMonths'] = df_model['CompetitionOpenMonths'].clip(lower=0)

df_model[['Store', 'Date', 'IsHoliday', 'CompetitionOpenMonths']].describe()
```

```python
# Cap unrealistic competition age (anything over 20 years is almost certainly a data artifact)
df_model['CompetitionOpenMonths'] = df_model['CompetitionOpenMonths'].clip(upper=240)

# Fill missing competition age with a sentinel (-1) indicating "unknown," and add a flag
df_model['CompetitionAgeUnknown'] = df_model['CompetitionOpenMonths'].isna().astype(int)
df_model['CompetitionOpenMonths'] = df_model['CompetitionOpenMonths'].fillna(-1)

# Fill missing CompetitionDistance with median (only 3 missing, low impact)
df_model['CompetitionDistance'] = df_model['CompetitionDistance'].fillna(df_model['CompetitionDistance'].median())

# One-hot encode categorical variables
df_model = pd.get_dummies(df_model, columns=['StoreType', 'Assortment'], drop_first=True)

print(df_model.shape)
df_model.columns.tolist()
```

```python
# One-hot encode StateHoliday
df_model = pd.get_dummies(df_model, columns=['StateHoliday'], drop_first=True)

# Drop columns we don't want in the model
drop_cols = ['Customers', 'Open', 'PromoInterval', 
             'CompetitionOpenSinceMonth', 'CompetitionOpenSinceYear',
             'Promo2SinceWeek', 'Promo2SinceYear']
df_model = df_model.drop(columns=drop_cols)

print(df_model.shape)
print(df_model.dtypes)
```

```python
# Time-based split: hold out the last 6 weeks as test set
split_date = df_model['Date'].max() - pd.Timedelta(weeks=6)

train_df = df_model[df_model['Date'] <= split_date].copy()
test_df = df_model[df_model['Date'] > split_date].copy()

print("Train date range:", train_df['Date'].min(), "to", train_df['Date'].max())
print("Test date range:", test_df['Date'].min(), "to", test_df['Date'].max())
print("Train shape:", train_df.shape)
print("Test shape:", test_df.shape)
```

```python
from prophet import Prophet

# Aggregate total sales per day across all stores
daily_sales = df_model.groupby('Date')['Sales'].sum().reset_index()
daily_sales.columns = ['ds', 'y']

# Split into train/test using the same cutoff date as before
train_daily = daily_sales[daily_sales['ds'] <= split_date].copy()
test_daily = daily_sales[daily_sales['ds'] > split_date].copy()

print(train_daily.shape, test_daily.shape)
train_daily.tail()
```

```python
# Fit Prophet model
model_prophet = Prophet(yearly_seasonality=True, weekly_seasonality=True, daily_seasonality=False)
model_prophet.fit(train_daily)

# Create future dataframe covering the test period
future = model_prophet.make_future_dataframe(periods=len(test_daily))

# Generate forecast
forecast = model_prophet.predict(future)

# Extract just the test period predictions
forecast_test = forecast[forecast['ds'] > split_date][['ds', 'yhat', 'yhat_lower', 'yhat_upper']]
forecast_test.head()
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
import numpy as np

# Merge actual vs predicted for the test period
eval_df = test_daily.merge(forecast_test[['ds', 'yhat']], on='ds')

mae = mean_absolute_error(eval_df['y'], eval_df['yhat'])
rmse = np.sqrt(mean_squared_error(eval_df['y'], eval_df['yhat']))
mape = np.mean(np.abs((eval_df['y'] - eval_df['yhat']) / eval_df['y'])) * 100

print(f"MAE: {mae:,.0f}")
print(f"RMSE: {rmse:,.0f}")
print(f"MAPE: {mape:.2f}%")
```

```python
from xgboost import XGBRegressor

# Define features (X) and target (y), excluding Date and Sales itself
feature_cols = [c for c in train_df.columns if c not in ['Date', 'Sales']]

X_train = train_df[feature_cols]
y_train = train_df['Sales']
X_test = test_df[feature_cols]
y_test = test_df['Sales']

print(X_train.shape, X_test.shape)
X_train.dtypes
```

```python
model_xgb = XGBRegressor(
    n_estimators=500,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42
)

model_xgb.fit(X_train, y_train)

# Predict on test set
y_pred = model_xgb.predict(X_test)

# Evaluate
mae_xgb = mean_absolute_error(y_test, y_pred)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred))
mape_xgb = np.mean(np.abs((y_test - y_pred) / y_test)) * 100

print(f"XGBoost MAE: {mae_xgb:,.0f}")
print(f"XGBoost RMSE: {rmse_xgb:,.0f}")
print(f"XGBoost MAPE: {mape_xgb:.2f}%")
```

```python
import matplotlib.pyplot as plt

# Feature importance
importance = pd.DataFrame({
    'feature': feature_cols,
    'importance': model_xgb.feature_importances_
}).sort_values('importance', ascending=False)

print(importance.head(15))

plt.figure(figsize=(10, 6))
plt.barh(importance['feature'][:15][::-1], importance['importance'][:15][::-1])
plt.xlabel('Importance')
plt.title('Top 15 Feature Importances - XGBoost')
plt.tight_layout()
plt.savefig('feature_importance.png')
plt.show()
```
